```python
import seaborn as sns
import matplotlib.pyplot as plt


data = sns.load_dataset('tips')


sns.scatterplot(x='total_bill', y='tip', data=data)

plt.title('Scatter Plot of Total Bill vs. Tip')
plt.xlabel('Total Bill')
plt.ylabel('Tip')

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = pd.read_csv('Desktop/Employee.csv')

plt.figure(figsize=(10, 6))
sns.scatterplot(data=data,
                x='Age',
                y='ExperienceInCurrentDomain')

plt.title('Age vs Experience in Current Domain')
plt.xlabel('Age')
plt.ylabel('Experience in Current Domain')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = pd.read_csv('Desktop/Employee.csv')

plt.figure(figsize=(10, 6))

sns.boxplot(x='Education', y='Age', data=data)

plt.title('Box Plot of Age by Education')
plt.xlabel('Education')
plt.ylabel('Age')

plt.show()
```
