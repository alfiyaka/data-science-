```python
rows = int(input("Enter the number of rows: "))

# Upper part
for i in range(1, rows + 1):
    print(" " * (rows - i) + "* " * i)

# Lower part
for i in range(rows - 1, 0, -1):
    print(" " * (rows - i) + "* " * i)
```

```python
rows = int(input("Enter the size: "))

for i in range(rows):
    for j in range(rows):
        if i == 0 or i == rows - 1 or j == 0 or j == rows - 1:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
```

```python
rows = int(input("Enter the number of rows: "))

for i in range(rows):
    ch = 'A'
    for j in range(i + 1):
        print(ch, end=" ")
        ch = chr(ord(ch) + 1)
    print()
```

```python
rows = int(input("Enter the number of rows: "))

for i in range(rows):
    print(" " * (rows - i), end="")
    num = 1
    for j in range(i + 1):
        print(num, end=" ")
        num = num * (i - j) // (j + 1)
    print()
```
